

### 6.3 Experiment-log template

```markdown
# Deep Learning Fundamentals — Experiment Log

## Environment
- Python: 3.13.16
- PyTorch:2.11.0+cpu
- Device:CPU
- Seed:42 (also 43 for the different-seed experiment)

## Dataset and preprocessing
- Dataset:MNIST handwritten digit dataset (digits 0–9).
- Train/validation/test split:55,000 training images, 5,000 validation images, and 10,000 test images.
- Input transformation:transforms.ToTensor() converts images into PyTorch tensors and scales pixel values from 0–255 to 0–1

## Model
- Architecture:CNN with two convolutional layers (8 and 16 filters), ReLU activation, two MaxPool2d layers, a Flatten layer, and a final Linear layer for classifying digits 0–9.

- Parameter count:9,098

## Training configuration
- Optimizer:SGD (Stochastic Gradient Descent)
- Learning rate:0.05
- Batch size:64
- Epochs:3

## Results
- Final training loss:0.0940 (seed 42), 0.0900 (seed 43)
- Final validation accuracy:96.44% (seed 42), 96.34% (seed 43)
- Final test accuracy:97.11% (Section 4)
- Relevant plots:Training Loss vs. Epoch (CNN Training Stability Across Random Seedst)

## Interpretation
- Evidence of underfitting/overfitting:No clear overfitting was seen in the 3 epochs
- Optimization stability:The training loss decreased steadily, showing that the model was learning well.
- Main comparison:Both runs with seed 42 gave the same validation accuracy of 96.44%. With seed 43, it was 96.34%, only 0.10 percentage points lower.
- Limitations:The model was trained for only 3 epochs using two different seeds, so the results may change with more training
```
